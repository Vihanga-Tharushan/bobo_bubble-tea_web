### Port Number Ranges and Categories

Port numbers, which range from $0$ to $65535$, are divided into three main categories by the Internet Assigned Numbers Authority (IANA) to organize network traffic and identify specific services.

1. **Well‑Known Ports ($0$–$1023$)**

- **Purpose:** Reserved for standard, foundational network services and protocols.
- **Privileges:** On Unix‑like systems these require superuser (root) privileges to bind.
- **Examples:** `20`/`21` — FTP, `22` — SSH, `25` — SMTP, `53` — DNS, `80` — HTTP, `443` — HTTPS.

2. **Registered Ports ($1024$–$49151$)**

- **Purpose:** Used by specific applications, software vendors, and services that register them with IANA to avoid conflicts.
- **Examples:** `1433`/`1434` — Microsoft SQL Server, `3306` — MySQL, `3389` — RDP, `8080` — HTTP (alternate).

3. **Dynamic / Private Ports ($49152$–$65535$)**

- **Also called:** Ephemeral or temporary ports.
- **Purpose:** Assigned temporarily by the operating system to client applications (for example, when a browser opens a connection).
- **Restrictions:** Not restricted—available for use by any application.

**Summary Table**

| Category | Range | Description |
|---|---:|---|
| Well‑Known | $0$–$1023$ | Standardized protocols, system services |
| Registered | $1024$–$49151$ | Vendor‑specific applications / software |
| Dynamic / Private | $49152$–$65535$ | Temporary client connections (ephemeral) |

## Project Scope (In Scope)

### 5.1 Customer Module

- User registration and login (Email / Google OAuth)
- Browse menu by categories (Bubble Tea, Cool Drinks, Ice Cream Sundae)
- View product details with images, descriptions, and pricing
- Customize drinks and sundaes:
  - Size: Small / Medium / Large
  - Sugar level: 0%, 25%, 50%, 75%, 100%
  - Ice level: No ice, Less ice, Normal, Extra ice
  - Toppings: Tapioca pearls, Jelly, Pudding, Whipped cream, and more
- Add items to cart with customization details
- Enter delivery address with area or zone validation
- Place orders with an order summary
- Use online payment integration (Stripe / PayHere for Sri Lanka)
- Choose Cash on Delivery (COD)
- Track order status in real time:
  - Order Placed → Preparing → Ready → Out for Delivery → Delivered
- View order history and reorder items
- Rate and review products
- Manage profile information
- Save favorite or wishlist products

### 5.2 Admin Module

- Secure admin login
- Dashboard with analytics:
  - Total orders, revenue, and top-selling items
  - Daily, weekly, and monthly reports
  - Charts and graphs using Chart.js or Recharts
- Menu management (CRUD):
  - Add, edit, and delete products
  - Set categories, prices, and available toppings
  - Upload product images
  - Mark items as available or unavailable
- Order management:
  - View all orders with filters for status, date, and area
  - Update order status
  - Cancel or refund orders
- Delivery zone management:
  - Define deliverable areas using pincode-based or radius-based validation
  - Set delivery charges per zone
  - Block or unblock delivery areas
- Customer management:
  - View customer list
  - View customer order history
- Promo code and discount management:
  - Create coupon codes
  - Set discount percentage or flat amount
  - Set validity period and usage limits
- Review and rating moderation

### 5.3 Delivery Module

- Delivery person registration and login
- View assigned orders
- Update delivery status (Picked Up → On the Way → Delivered)
- View delivery history
- Contact customer using phone number visibility

### 5.4 Delivery Zone Validation

- Validate customer address against serviceable areas during checkout
- Option 1: Pincode or area dropdown-based validation
- Option 2: Google Maps API radius-based validation
- Display estimated delivery time based on zone
- Show delivery charge based on distance or zone
- Show a clear rejection message for non-deliverable areas

# Database Collections

### Users

- **Fields:** name, email, password, phone
- **Role:** "customer" | "admin" | "delivery"
- **Favorites:** [productId]
- **Addresses:** [{ address, city, pincode, coordinates }]

### Products

- **Fields:** name, description, category, image
- **Pricing:** basePrice, sizes: [{ name, price }]
- **Toppings:** availableToppings: [{ name, price }]
- **Options:** sugarLevels, iceLevels
- **Meta:** isAvailable, ratings, numReviews, createdAt

### Orders

- **userId**
- **items:** [{ productId, customization, qty, price }]
- **Delivery:** deliveryAddress, deliveryZone, deliveryCharge, estimatedDeliveryTime
- **Payment:** paymentMethod, paymentStatus, totalAmount
- **Status:** orderStatus, deliveryPersonId, timestamps

### Delivery Zones

- **zoneName**
- **pincodes:** []
- **Settings:** deliveryCharge, estimatedTime, isActive

### Reviews

- **userId**, **productId**
- **rating**, **comment**, **createdAt**

### Promo Codes

- **code**, **discountType**, **discountValue**
- **Limits:** minOrderAmount, maxDiscount, usageLimit
- **Validity:** validFrom, validTo, isActive
